### Amostragem sitemática
Seleciona um elemento a partir da posição $r$ a cada intervalo $k$, onde $k$ é dado por:
$$ k = \left\lceil\frac{N}{n} \right\rceil$$
sendo $N$ o número de elementos da população e $n$ o número de elementos que serão selecionados para a amostra

```python
import pandas as pd
import numpy as np
from math import ceil
```

```python
N = 150
n = 15
k = ceil(N / n)
print(k)
```

```python
r = np.random.randint(low= 1, high = k+1, size=1) # Seleciona um inteiro aletório no intervalo [1, k+1]
print(r)
```

```python
acumulador = int(r[0])
sorteados = []
for i in range(n):
  sorteados.append(acumulador)
  acumulador += k
print(sorteados)
# Sorteados - Array com os indices dos elementos que serão sorteados
```

```python
base = pd.read_csv('iris.csv')
base_final = base.loc[sorteados]
base_final
```

```python
import pandas as pd
import numpy as np
import matplotlib as plt
```

```python
base = pd.read_csv('iris.csv');
base.shape
```

```python
rng = np.random.default_rng()

# a - Valores que serão escolhidos
# p - Probabilidade de um valor ser escolhido, respectivamente
# replace - Com reposição
# size - Tamanho da amostra
amostra = rng.choice(a=[0,1],  size=150, p=[0.3, 0.7], replace=True)

len(amostra[amostra == 0])
```

```python
base_final = base.loc[amostra == 0]; # Compara com a base de dados inicial, os elementos com zero na amostra são selecionados
base_final.shape
```

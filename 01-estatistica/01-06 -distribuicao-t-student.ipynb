# Distribuição T de Student

Usada especialmente para estimar a média de uma população quando o **tamanho da amostra** é pequeno ($n \lt 30$) e **desvio padrão populacional** ($\sigma$) é desconhecido

## Graus de liberdade

Definem o formato da curva. Quanto menores os graus de liberdade, mais achata é a curva

são comumente definidos por $df = n - 1$

## Fórmula estatísitica
$$t = \frac{\bar{X} - \mu_0}{S/\sqrt{n}}$$

```python
from scipy.stats import t
```

- Média($\bar{X}$): 75
- Amostra($n$) = 9
- Desvio padrão($S$) = 10

```python
# Qual a probabilidade de selecionar menor 80?
#     t    grau de liberdade
t.cdf(1.5, 8) 
```

```python
# Qual a probabilidade de ser maior que 80
t.sf(1.5, 8)
```

## Distribuição Normal (Gaussiana)

A **Distribuição Normal** é uma distribuição de probabilidade contínua fundamental na estatística e na ciência de dados, caracterizada por sua curva simétrica em formato de sino. Nela, a maioria das observações se concentra em torno da média, enquanto valores muito distantes (outliers) tornam-se progressivamente mais raros.

Ela é inteiramente definida por dois parâmetros:
* **Média ($\mu$):** determina o centro (pico) da distribuição.
* **Desvio Padrão ($\sigma$):** define a dispersão ou "largura" da curva em torno da média.

### Função Densidade de Probabilidade (FDP)
$$f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x - \mu}{\sigma}\right)^2}$$


```python
from scipy.stats import norm
```

### Função Distribuição Acumulada(Cumulative Distribuition Function)
O método `norm.cdf` calcula a área sob à curva à esquerda de um valor $x$
$$\text{CDF}(X) = P(X \le x) = \int_{-\infty}^{x} f(t)\,dt$$


```python
# Conjunto de objetos com pesos normalmente distribuídos, média 8 e desvio padrão 2
# Qual a prob de tira um objeto que pesa menos que 6?
norm.cdf(6, 8, 2)
```

```python
# Qual a prob de tirar um objeto que pesa mais que 6?
p = 1 - norm.cdf(6, 8, 2)
p
```

```python
# Qual a prob de tirar um objeto que pesa mais que 8 e menos que 10?
p = norm.cdf(10, 8, 2) - norm.cdf(8, 8, 2)
p
```

### Função de sobrevivência(Surivival Function)
O método `norm.sf` representa a probabilidade de uma variável aleatória assumir um valor maior que um determinado ponto $x$
$$
  \text{SF}(x) = P(X \gt x) = 1 - \text{CDF}(x)
$$

```python
# Qual a prob de tirar um objeto que pesa menos de 6 e mais que 10 quilos
p1 = norm.cdf(6, 8, 2) + norm.sf(10, 8, 2)
p1
```

```python
# Qual a prob de tirar um objeto que pesa mais que 8 e menos que 10?
p = norm.sf(8, 8, 2) - norm.sf(10, 8, 2)
p
```
